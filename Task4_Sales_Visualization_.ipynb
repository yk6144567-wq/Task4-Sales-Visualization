# Task 4: Sales Data Visualization & Storytelling

## Project Overview

This project analyzes monthly sales, profit, and order data using Python. The purpose is to identify business trends, compare monthly performance, and communicate useful insights through charts.

## Objectives

* Analyze monthly sales and profit.
* Identify the highest-performing month.
* Visualize trends using charts.
* Provide data-driven business recommendations.

## Tools Used

* Python
* Pandas
* Matplotlib
* Google Colab

## Dataset Description

The dataset contains 12 months of sample data with four columns: Month, Sales, Profit, and Orders.

**Note:** This is illustrative practice data, not verified real-company data.


```python

import pandas as pd

data = {
    "Month": [
        "January", "February", "March", "April",
        "May", "June", "July", "August",
        "September", "October", "November", "December"
    ],
    "Sales": [
        12000, 15000, 14000, 18000,
        22000, 21000, 25000, 27000,
        24000, 30000, 35000, 40000
    ],
    "Profit": [
        2000, 2800, 2500, 3500,
        4500, 4000, 5200, 6000,
        4800, 7000, 8500, 10000
    ],
    "Orders": [
        120, 140, 135, 160, 190, 180,
        210, 225, 200, 250, 280, 320
    ]
}

df = pd.DataFrame(data)

print("SALES DATASET")
display(df)

print("Dataset Shape:", df.shape)
```

```python

# STEP 3: DATA ANALYSIS

# 1. Total Sales, Profit and Orders
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Orders"].sum()

print("===== BUSINESS SUMMARY =====")
print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Total Orders:", total_orders)

# 2. Highest Sales Month
best_sales_month = df.loc[df["Sales"].idxmax()]

print("\n===== HIGHEST SALES MONTH =====")
print("Month:", best_sales_month["Month"])
print("Sales:", best_sales_month["Sales"])

# 3. Highest Profit Month
best_profit_month = df.loc[df["Profit"].idxmax()]

print("\n===== HIGHEST PROFIT MONTH =====")
print("Month:", best_profit_month["Month"])
print("Profit:", best_profit_month["Profit"])

# 4. Profit Margin
profit_margin = (total_profit / total_sales) * 100

print("\n===== PROFIT MARGIN =====")
print(round(profit_margin, 2), "%")

# 5. Monthly Average
print("\nAverage Monthly Sales:", round(df["Sales"].mean(), 2))
print("Average Monthly Profit:", round(df["Profit"].mean(), 2))
```

```python

import matplotlib.pyplot as plt

# Chart 1: Monthly Sales Trend
plt.figure(figsize=(10, 5))
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()

# Chart 2: Monthly Profit Comparison
plt.figure(figsize=(10, 5))
plt.bar(df["Month"], df["Profit"])
plt.title("Monthly Profit Comparison")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Chart 3: Sales vs Profit
plt.figure(figsize=(8, 5))
plt.scatter(df["Sales"], df["Profit"])
plt.title("Sales vs Profit Analysis")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

# STEP 5: BUSINESS STORYTELLING

print("CHART 1: MONTHLY SALES TREND")
print("Takeaway: Sales increased from Rs. 12,000 in January to Rs. 40,000 in December.")
print("Business Insight: Sales show an overall upward trend during the year.")

print("\nCHART 2: MONTHLY PROFIT COMPARISON")
print("Takeaway: December recorded the highest profit of Rs. 10,000.")
print("Business Insight: The business should investigate what drove December's strong profit.")

print("\nCHART 3: SALES VS PROFIT")
print("Takeaway: Higher sales are generally associated with higher profit in this sample.")
print("Business Insight: Monitor costs and profit margins while growing sales.")

print("\nFINAL BUSINESS RECOMMENDATION")
print("Study the strategies used in high-performing months.")
print("Plan inventory and marketing around customer demand.")
print("Track monthly profit margins, not just sales.")
```

```python

# STEP 6: PROFESSIONAL VISUAL STORY REPORT
# No downloads required

import matplotlib.pyplot as plt
from IPython.display import display, HTML

# Calculate key metrics from existing dataset
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Orders"].sum()
profit_margin = total_profit / total_sales * 100

best_sales = df.loc[df["Sales"].idxmax()]
best_profit = df.loc[df["Profit"].idxmax()]

# Report styling
style = """
<style>
.story-page {
    font-family: Arial, sans-serif;
    padding: 28px;
    margin: 18px auto;
    max-width: 850px;
    border: 1px solid #d9e2ec;
    border-radius: 12px;
    background: white;
    color: #243447;
    line-height: 1.6;
}
.story-page h1, .story-page h2 { color: #174a7e; }
.story-page .metric {
    display: inline-block;
    padding: 12px;
    margin: 5px;
    background: #edf5ff;
    border-radius: 8px;
    min-width: 130px;
}
.story-page .note { color: #526579; font-size: 13px; }
</style>
"""

def page(title, content):
    display(HTML(
        style + '<section class="story-page">'
        + '<h2>' + title + '</h2>'
        + content + '</section>'
    ))

# PAGE 1 — Cover
page(
    "01 | SALES DATA VISUALIZATION",
    "<h1>Sales Data Visualization & Storytelling</h1>"
    "<p><b>Project:</b> Task 4 — Data Analytics</p>"
    "<p>This report explores monthly sales, profit and order trends "
    "using a sample dataset and Python visualizations.</p>"
    "<p><b>Prepared by:</b> Yash Kumar</p>"
    "<p class='note'>Practice project using illustrative data.</p>"
)

# PAGE 2 — Executive summary
page(
    "02 | EXECUTIVE SUMMARY",
    f"<div class='metric'><b>Total Sales</b><br>Rs. {total_sales:,.0f}</div>"
    f"<div class='metric'><b>Total Profit</b><br>Rs. {total_profit:,.0f}</div>"
    f"<div class='metric'><b>Total Orders</b><br>{total_orders:,}</div>"
    f"<div class='metric'><b>Profit Margin</b><br>{profit_margin:.2f}%</div>"
    "<p>Sales increased overall across the sample year. "
    "December was the strongest month for both sales and profit.</p>"
)

# PAGE 3 — Sales trend
page(
    "03 | MONTHLY SALES TREND",
    f"<p>Sales rose from Rs. {df.iloc[0]['Sales']:,.0f} in "
    f"{df.iloc[0]['Month']} to Rs. {best_sales['Sales']:,.0f} in "
    f"{best_sales['Month']}.</p>"
    "<p><b>Takeaway:</b> The dataset shows an overall upward sales trend.</p>"
)
plt.figure(figsize=(10, 4))
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales (Rs.)")
plt.xticks(rotation=40)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# PAGE 4 — Profit performance
page(
    "04 | MONTHLY PROFIT",
    f"<p>Highest monthly profit: Rs. {best_profit['Profit']:,.0f} "
    f"in {best_profit['Month']}.</p>"
    "<p><b>Takeaway:</b> Compare costs, pricing and demand in high-profit "
    "months to understand what contributed to the result.</p>"
)
plt.figure(figsize=(10, 4))
plt.bar(df["Month"], df["Profit"])
plt.title("Monthly Profit Comparison")
plt.xlabel("Month")
plt.ylabel("Profit (Rs.)")
plt.xticks(rotation=40)
plt.tight_layout()
plt.show()

# PAGE 5 — Sales versus profit
page(
    "05 | SALES VS PROFIT",
    "<p>The scatter plot helps assess how sales and profit move together.</p>"
    "<p><b>Takeaway:</b> Higher sales are generally associated with higher "
    "profit in this sample. This does not prove that sales alone cause "
    "profit to increase.</p>"
)
plt.figure(figsize=(7, 4))
plt.scatter(df["Sales"], df["Profit"])
plt.title("Sales vs Profit")
plt.xlabel("Sales (Rs.)")
plt.ylabel("Profit (Rs.)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# PAGE 6 — Recommendations and conclusion
page(
    "06 | RECOMMENDATIONS & CONCLUSION",
    "<ol>"
    "<li>Investigate the strategies used in high-performing months.</li>"
    "<li>Monitor costs and profit margins alongside sales.</li>"
    "<li>Use demand trends to plan inventory and marketing.</li>"
    "<li>Collect product, region and expense data for deeper analysis.</li>"
    "</ol>"
    "<p><b>Conclusion:</b> The sample dataset demonstrates how "
    "visualizations can communicate sales and profit trends to support "
    "business decisions.</p>"
    "<p class='note'>All figures are based on practice data and should not "
    "be treated as verified results from a real business.</p>"
)

print("Report displayed above. Scroll up to review all six sections.")
```
